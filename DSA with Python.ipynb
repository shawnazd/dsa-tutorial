# Time Complexity

Time Complexity measures how long an algorithm takes to run as the size of the input grows. It helps us analyze the efficiency of an algorithm and predict its performance on large datasets.

Time complexity is usually expressed using Big-O notation, such as O(1), O(n), and O(n²).

# Big-O notation

## O(1) – Constant Time Complexity

An algorithm has O(1) time complexity if the execution time does not change with input size.

Example: Accessing a specific element in an array.

```
arr = [10, 20, 30, 40]
print(arr[0])  # Always takes the same time
```

In this case, the operation takes the same amount of time whether the array contains 4 elements or 1,000 elements.

## O(n) – Linear Time Complexity

An algorithm has O(n) time complexity if the execution time increases proportionally with the input size.

Example: Traversing all elements in an array.


```
arr = [10, 20, 30, 40]
for x in arr:
    print(x)
```
If the array has n elements, the loop will execute n times, so the running time grows proportionally with n.

## O(n²) – Quadratic Time Complexity

An algorithm has O(n²) time complexity if the execution time grows with the square of the input size. This often happens when a loop is inside another loop.
Example:

```
for i in range(n):
    for j in range(n):
        print(i, j)
```

If n = 10, the total number of operations is approximately 10 × 10 = 100.

## O(n log n) – Log-Linear Time Complexity

An algorithm has O(n log n) time complexity if the execution time grows as n multiplied by log n. This usually happens when the problem is repeatedly divided into smaller parts and then processed, like in divide-and-conquer algorithms such as Quicksort.

Example:

```
def example(n):
    i = 1
    while i < n:
        for j in range(n):
            print(j)
        i *= 2
```
Why it’s O(n log n):
Outer loop (while i < n)
i doubles each time: 1, 2, 4, 8…
Runs ≈ log₂(n) times
Inner loop (for j in range(n))
Runs n times for each outer loop iteration
Total work = inner × outer = n × log n

Example: if n = 10 → 10 × log₂10 ≈ 33 operations

# Best, Average, and Worst Case

* **Best Case (Ω):** Runs fastest; input is ideal.
* **Average Case (θ):** Runs typical/expected time; realistic scenario.
* **Worst Case (O):** Runs slowest; input is worst-case.

---

Table – Linear Search Example

| Case    | Symbol | Meaning               | Example                  |
| ------- | ------ | --------------------- | ------------------------ |
| Best    | Ω      | Minimum time          | Target is first element  |
| Average | θ      | Typical/expected time | Target is middle element |
| Worst   | O      | Maximum time          | Target not in array      |

---

**Python Example:**

```python
arr = [10, 20, 30, 40]
target = 10

for x in arr:
    if x == target:
        break  # Best: first, Average: middle, Worst: not present
```

# Space Complexity

Space complexity measures **how much memory an algorithm needs** to run as the input size increases.

It has **two main parts**:

1. **Input Space** – Memory used to store the **input itself**.

   * Example: An array of size n → uses **O(n)** memory.

2. **Auxiliary Space** – Extra memory used by the algorithm (not counting input).

   * Example: Variables, temporary arrays, recursion stack.

---

Examples

1. **O(1) – Constant Space**

* Uses fixed memory regardless of input size.

```python
x = 5
y = 10
z = x + y
```

2. **O(n) – Linear Space**

* Memory grows linearly with input size.

```python
arr = [0] * n  # Array of size n
```

3. **O(n²) – Quadratic Space**

* Memory grows with the square of input size.

```python
matrix = [[0]*n for _ in range(n)]  # n x n matrix
```

---

Summary Table

| Space Complexity | Meaning          | Example         |
| ---------------- | ---------------- | --------------- |
| O(1)             | Constant memory  | Few variables   |
| O(n)             | Linear memory    | Array of size n |
| O(n²)            | Quadratic memory | n x n matrix    |

# Time Limit Exceeded (TLE) Error

**TLE (Time Limit Exceeded)**

TLE occurs when a program does not complete execution within the allowed time. It usually happens when the algorithm is too slow for the input size. Roughly, one second can handle about 10⁸ simple operations in C++, less in Python.

For example, in Python, a loop running a billion times like

```python
for i in range(10**9):
    pass
```

will cause TLE because it exceeds the typical time limit of one second. Using a smaller loop, such as

```python
for i in range(10**5):
    pass
```

will usually run within the time limit.

Efficient algorithms with lower time complexity prevent TLE, especially for large inputs.

# Hashing

````python
# Brute Force Solution
n = [5, 3, 2, 2, 1, 5, 5, 7, 5, 10]
m = [10, 111, 1, 9, 5, 67, 2]

for num in m:
  count = 0
  for x in n:
    if x == num:
      count += 1
  print(f"{num}: {count}")
````

````python
# Hash List Solution
n = [5, 3, 2, 2, 1, 5, 5, 7, 5, 10]
m = [10, 111, 1, 9, 5, 67, 2]

hash_list = [0] * 11

for num in n:
    hash_list[num] += 1

for x in m:
    if 0 <= x < len(hash_list):
        print(f"{x}: {hash_list[x]}")
    else:
        print(f"{x}: 0")
````

````python
# Using Dictionary
n = [5, 3, 2, 2, 1, 5, 5, 7, 5, 10]
m = [10, 111, 1, 9, 5, 67, 2]

hash_dict = {}

for i in n:
    hash_dict[i] = hash_dict.get(i, 0) + 1

for x in m:
    print(f"{x}: {hash_dict.get(x, 0)}")
````

````python
# Character Hashing
s = "azyxyyzaaaa"
q = ["c", "a", "y", "x"]

hash_list = [0] * 26  # 26 letters

for i in s:
    index = ord(i) - 97
    hash_list[index] += 1

for ch in q:
    index = ord(ch) - 97
    print(f"{ch}: {hash_list[index]}")
````

````python
# Character Hashing
s = "aZyXyyZaAAA!@#"
q = ["c", "a", "y", "x", "Z", "!", "@"]

hash_dict = {}

for ch in s:
    hash_dict[ch] = hash_dict.get(ch, 0) + 1

for ch in q:
    print(f"{ch}: {hash_dict.get(ch, 0)}")
````

# Recursion

A technique where a function calls itself to solve smaller instances of a problem until a base case stops it.

````python
def print_numbers(n):
    if n == 0:
        return
    print(n)
    print_numbers(n-1)

print_numbers(50)
````

````python
# print 1 to N using recursion
def print_numbers(n):
  if n == 0:
    return
    print_numbers(n - 1)
    print()

print_numbers(10)
````

````python
# print N to 1 using recursion
def print_numbers(n):
    if n == 0:
        return
    print(n)
    print_numbers(n-1)

print_numbers(10)
````

## Fibonacci Number

````python
def fibonacci(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)

# Example usage
print(fibonacci(6))  # Output: 8
````

````python
def fibonacci(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(5))
````

````python
def fibonacci(n):
  a, b = 0, 1
  for _ in range(n):
    a, b = b, a + b
  print(a)

fibonacci(5)
````
